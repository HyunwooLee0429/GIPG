# GKG driver: generate Duguet-style instances

This notebook lives in `src/gkg/`. It writes JSON instances to the project-root folder `data/gkg_instances/`.


```python
from pathlib import Path
import sys

# Ensure we can import the `gkg` package when running from src/gkg/
    

```

```python
import json
from gipg.gkg.instance import generate_gkg_instance

# Output folder (project-root/data/gkg_instances)
OUTDIR = Path('../data/gkg') / 'gkg_instances'
OUTDIR.mkdir(parents=True, exist_ok=True)

# Generator parameters (Duguet et al. grid)
seed0 = 12345
R = 1000
ns = [2, 3, 4]
ms = [5, 10, 15, 20, 30, 40, 50]
cap_factors = [0.2, 0.5]
corrs = ['uncorrelated', 'weakly_correlated', 'strongly_correlated']
reps = 10

k = 0
for n in ns:
    for m in ms:
        for cf in cap_factors:
            for corr in corrs:
                for r in range(reps):
                    seed = seed0 + k * 1009 + r * 9176
                    inst = generate_gkg_instance(n=n, m=m, cap_factor=cf, corr=corr, seed=seed, R=R)
                    fname = f'gkg_n{n}_m{m}_cf{cf}_corr{corr}_rep{r}.json'
                    with open(OUTDIR / fname, 'w', encoding='utf-8') as f:
                        json.dump(inst.to_dict(), f)
                    k += 1

print(f'Wrote {k} instances to: {OUTDIR}')
```

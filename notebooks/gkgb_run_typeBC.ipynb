# Bilinear KPG — Type BC (Semi-Negative) Experiments

**New interaction type**: Type BC with $C_{ij} \sim U[-50, 100]$ (between Type B and Type C).  
- Type B (Dragotto): $C_{ij} \sim U[1, 100]$ — purely positive interactions  
- Type C (Dragotto): $C_{ij} \sim U[-100, 100]$ — fully mixed interactions  
- **Type BC (ours)**: $C_{ij} \sim U[-50, 100]$ — skewed positive with some negative  

**Dataset**: 24 generated instances, n ∈ {2, 3}, m ∈ {25, 50, 75, 100}, budget_ratio ∈ {2, 5, 8}  
- Shared item capacities $c_j \sim U\{1, n\}$ (seed=42), same as Dragotto experiments  

**Exp 1**: BRD + GZR @ α=1 → Best PNE, POS  
**Exp 2**: TAS for instances where Exp 1 failed to find a PNE

## 1. Load Type BC Instances

```python
# Path to Type BC instances
DATA_DIR = Path('../data/gkgb/type_BC')
assert DATA_DIR.exists(), f'Missing data folder: {DATA_DIR}'

SHARED_SEED = 42

instances = enumerate_dragotto_instances(DATA_DIR, types=('BC',), shared_seed=SHARED_SEED)
print(f'Loaded {len(instances)} Type BC instances from {DATA_DIR}')

# Summary
print('\n--- By (n, m) ---')
size_counts = Counter()
for inst in instances:
    size_counts[(inst.n, inst.m)] += 1
for (n, m), cnt in sorted(size_counts.items()):
    print(f'  n={n}, m={m}: {cnt} instances')

print(f'\n--- Interaction coefficient range ---')
all_C = np.concatenate([inst.C.flatten() for inst in instances])
nonzero_C = all_C[all_C != 0]
print(f'  C range (nonzero): [{nonzero_C.min()}, {nonzero_C.max()}]')
print(f'  C mean (nonzero): {nonzero_C.mean():.1f}')
print(f'  Fraction negative: {(nonzero_C < 0).mean():.1%}')
```

## 2. Experiment 1: Best PNE and POS (BRD + GZR @ α=1)

For each instance:
1. **BRD** warm-starts **GZR @ α=1** (if BRD found a PNE)
2. **GZR** with `stop_at_first=False` — optimises to MIP gap=0, finding the **best** PNE
3. **Social Optimum** computed separately
4. **POS** = SO / best_PNE

```python
TOTAL_TIME_LIMIT = 600.0  # Total budget for BRD + GZR combined
SO_TIME_LIMIT = 600.0     # 10 minutes for social optimum
BR_TIME_LIMIT = 10.0

results_exp1 = []

for idx, inst in enumerate(instances):
    tag = inst.tag
    parts = tag.split('-')
    n_val, m_val, br_val = int(parts[0]), int(parts[1]), int(parts[2])

    row = {
        'tag': tag,
        'n': inst.n,
        'm': inst.m,
        'budget_ratio': br_val,
        'interaction_type': inst.interaction_type,
    }

    # --- Phase 1: BRD ---
    try:
        x_brd, brd_pne, brd_time = brd_random_restart(
            inst, max_init=3, max_round=15, seed=0,
            br_time_limit=BR_TIME_LIMIT,
        )
        row['brd_found_pne'] = brd_pne
        row['brd_time'] = brd_time
        row['brd_alpha'] = 1.0 if brd_pne else alpha_of_profile(inst, x_brd, br_time_limit=BR_TIME_LIMIT)
        row['initial_pne_welfare'] = social_welfare(inst, x_brd)
    except Exception as e:
        row['brd_found_pne'] = False
        row['brd_time'] = 0.0
        row['brd_alpha'] = float('inf')
        row['initial_pne_welfare'] = None
        x_brd = None
        brd_pne = False
        brd_time = 0.0
        print(f'  BRD ERROR: {e}')

    # --- Phase 2: GZR @ alpha=1, warm-started with BRD profile ---
    gzr_time_limit = max(TOTAL_TIME_LIMIT - brd_time, 60.0)
    warm = x_brd if brd_pne else None

    try:
        gzr_res = solve_gzr(
            inst, alpha=1.0, time_limit=gzr_time_limit,
            br_time_limit=BR_TIME_LIMIT,
            warm_start=warm,
            stop_at_first=False, verbose=False,
        )
        row['gzr_status'] = gzr_res.status
        row['gzr_mip_gap'] = gzr_res.mip_gap
        row['gzr_obj_bound'] = gzr_res.obj_bound
        row['gzr_cuts'] = gzr_res.cuts_added
        row['gzr_br_calls'] = gzr_res.br_calls
        row['gzr_time'] = brd_time + gzr_res.runtime
        row['gzr_obj_val'] = gzr_res.obj_val
        row['gzr_first_pne_time'] = (
            brd_time + gzr_res.first_pne_time
            if gzr_res.first_pne_time is not None else None
        )
    except Exception as e:
        gzr_res = None
        row['gzr_status'] = 'ERROR'
        row['gzr_mip_gap'] = None
        row['gzr_obj_bound'] = None
        row['gzr_cuts'] = None
        row['gzr_br_calls'] = None
        row['gzr_time'] = None
        row['gzr_obj_val'] = None
        row['gzr_first_pne_time'] = None
        print(f'  GZR ERROR: {e}')

    # Best PNE welfare
    if gzr_res is not None and gzr_res.profile is not None and gzr_res.status in ('OPTIMAL', 'FEASIBLE'):
        row['best_pne_welfare'] = social_welfare(inst, gzr_res.profile)
    elif brd_pne:
        row['best_pne_welfare'] = row.get('initial_pne_welfare')
    else:
        row['best_pne_welfare'] = None

    row['first_pne_time'] = row.get('gzr_first_pne_time')

    # --- Phase 3: Social Optimum ---
    try:
        so_res = solve_social_optimum(inst, time_limit=SO_TIME_LIMIT, verbose=False)
        row['so_status'] = so_res.status
        row['so_welfare'] = so_res.opt_cost
        row['so_time'] = so_res.runtime
    except Exception as e:
        row['so_status'] = 'ERROR'
        row['so_welfare'] = None
        row['so_time'] = None
        print(f'  SO ERROR: {e}')

    # --- POS ---
    row['pos'] = None
    if (row.get('best_pne_welfare') is not None
            and row.get('so_welfare') is not None
            and row['best_pne_welfare'] > 0):
        row['pos'] = compute_pos(row['so_welfare'], row['best_pne_welfare'])

    results_exp1.append(row)

    pne_str = 'BRD-PNE' if brd_pne else 'no-PNE'
    gzr_time_str = f'{row["gzr_time"]:.1f}s' if row.get('gzr_time') is not None else '?s'
    gzr_cuts_str = f'{row["gzr_cuts"]}cuts' if row.get('gzr_cuts') is not None else '?cuts'
    pos_str = f'POS={row["pos"]:.3f}' if row['pos'] is not None else 'POS=N/A'
    fpne_str = f'1stPNE={row["gzr_first_pne_time"]:.1f}s' if row.get('gzr_first_pne_time') is not None else '1stPNE=N/A'
    print(f'[{idx+1}/{len(instances)}] {tag}: {pne_str} | GZR {row.get("gzr_status","?")} '
          f'({gzr_time_str}, {gzr_cuts_str}) | {pos_str} | {fpne_str}')

df_exp1 = pd.DataFrame(results_exp1)
df_exp1.to_csv(RESULTS_DIR / 'kpg_bilinear_typeBC_exp1_brd_gzr_pos.csv', index=False)
print(f'\nSaved {len(df_exp1)} rows to kpg_bilinear_typeBC_exp1_brd_gzr_pos.csv')
```

```python
# Experiment 1 Summary
print('=== Experiment 1 Summary (Type BC) ===')

n_total = len(df_exp1)
n_brd_pne = df_exp1['brd_found_pne'].sum()
n_gzr_opt = (df_exp1['gzr_status'] == 'OPTIMAL').sum()
n_gzr_feas = (df_exp1['gzr_status'] == 'FEASIBLE').sum()
n_gzr_inf = (df_exp1['gzr_status'] == 'INFEASIBLE').sum()
n_gzr_tl = (df_exp1['gzr_status'] == 'TIME_LIMIT').sum()
n_pos = df_exp1['pos'].notna().sum()

print(f'Total instances: {n_total}')
print(f'BRD found PNE: {n_brd_pne} ({n_brd_pne/n_total:.1%})')
print(f'GZR OPTIMAL: {n_gzr_opt} ({n_gzr_opt/n_total:.1%})')
print(f'GZR FEASIBLE: {n_gzr_feas}')
print(f'GZR INFEASIBLE: {n_gzr_inf}')
print(f'GZR TIME_LIMIT: {n_gzr_tl}')
print(f'POS available: {n_pos}')

# GZR found = BRD found PNE OR GZR OPTIMAL
gzr_found = ((df_exp1['brd_found_pne']) | (df_exp1['gzr_status'] == 'OPTIMAL')).sum()
print(f'GZR found PNE (BRD∪GZR): {gzr_found} ({gzr_found/n_total:.1%})')

if n_pos > 0:
    pos_vals = df_exp1['pos'].dropna()
    print(f'\nPOS stats: mean={pos_vals.mean():.4f}, max={pos_vals.max():.4f}, min={pos_vals.min():.4f}')

# Comparison with Type B and C
print('\n--- Quick comparison with Type B and C ---')
exp1_bc_path = RESULTS_DIR / 'kpg_bilinear_exp1_brd_gzr_pos.csv'
if exp1_bc_path.exists():
    df_bc = pd.read_csv(exp1_bc_path)
    for t in ['B', 'C']:
        dt = df_bc[df_bc['interaction_type'] == t]
        n_found = ((dt['brd_found_pne']) | (dt['gzr_status'] == 'OPTIMAL')).sum()
        pos_m = dt['pos'].dropna().mean() if dt['pos'].notna().sum() > 0 else float('nan')
        print(f'  Type {t}: {len(dt)} inst, GZR found={n_found}, mean POS={pos_m:.4f}')
    print(f'  Type BC: {n_total} inst, GZR found={gzr_found}, mean POS={pos_vals.mean():.4f}' if n_pos > 0 else f'  Type BC: {n_total} inst, GZR found={gzr_found}')
```

## 3. Experiment 2: Tightest Alpha Search (TAS) for Non-PNE Instances

Only instances where Exp 1 GZR reported **INFEASIBLE** or **TIME_LIMIT** (and BRD found no PNE).  
Bisection with verified/unverified bounds.

```python
# Load Exp1 results (use this cell if restarting kernel)
df_exp1 = pd.read_csv(RESULTS_DIR / 'kpg_bilinear_typeBC_exp1_brd_gzr_pos.csv')
results_exp1 = df_exp1.to_dict('records')
print(f'Loaded {len(df_exp1)} Exp1 rows')
```

```python
BR_TIME_LIMIT = 10.0
TAS_GZR_LIMIT = 1800.0   # 30 minutes per GZR call in TAS
TAS_EPS = 1e-2
TAS_MAX_ITERS = 30

# Gather failed instances from Exp1
failed_indices = [
    idx for idx, r in enumerate(results_exp1)
    if r['gzr_status'] in ('INFEASIBLE', 'TIME_LIMIT') and not r.get('brd_found_pne', False)
]
print(f'Exp2 candidates: {len(failed_indices)} instances')

results_exp2 = []

for count, idx in enumerate(failed_indices):
    inst = instances[idx]
    tag = inst.tag
    start_time = time.time()

    row = {
        'tag': tag,
        'n': inst.n,
        'm': inst.m,
        'budget_ratio': int(tag.split('-')[2]),
        'interaction_type': inst.interaction_type,
    }

    # --- Phase 0: BRD upper bound on alpha ---
    x_brd, brd_pne, _ = brd_random_restart(
        inst, max_init=5, max_round=20, seed=0,
        br_time_limit=BR_TIME_LIMIT,
    )
    alpha_init = 1.0 if brd_pne else alpha_of_profile(inst, x_brd, br_time_limit=BR_TIME_LIMIT)
    ub_verified = alpha_init
    best_profile = x_brd

    # --- Phase 1: Reuse Exp1 result for alpha=1 ---
    exp1_row = results_exp1[idx]
    if exp1_row['gzr_status'] == 'INFEASIBLE':
        lb_verified = 1.0
        lb_unverified = 1.0
    else:  # TIME_LIMIT
        lb_verified = None
        lb_unverified = 1.0

    # --- Phase 2: Bisection ---
    n_iters = 0
    history = []

    while (ub_verified - lb_unverified > TAS_EPS) and (n_iters < TAS_MAX_ITERS):
        n_iters += 1
        alpha_mid = 0.5 * (lb_unverified + ub_verified)

        gzr_mid = solve_gzr(
            inst, alpha=alpha_mid, time_limit=TAS_GZR_LIMIT,
            br_time_limit=BR_TIME_LIMIT,
            stop_at_first=True, verbose=False,
        )

        if gzr_mid.status in ('FEASIBLE', 'OPTIMAL') and gzr_mid.profile is not None:
            ub_verified = alpha_mid
            best_profile = gzr_mid.profile
            bound_type = 'verified_ub'
        elif gzr_mid.status == 'INFEASIBLE':
            lb_unverified = alpha_mid
            lb_verified = alpha_mid if lb_verified is None else max(lb_verified, alpha_mid)
            bound_type = 'verified_lb'
        else:
            lb_unverified = alpha_mid
            bound_type = 'unverified_lb'

        history.append({
            'iter': n_iters, 'alpha_mid': alpha_mid,
            'status': gzr_mid.status, 'bound_type': bound_type,
            'runtime': gzr_mid.runtime,
        })
        print(f'  iter {n_iters}: alpha={alpha_mid:.4f} -> {gzr_mid.status} ({bound_type}) '
              f'[{gzr_mid.runtime:.1f}s]')

    verified = (lb_verified is not None) and (ub_verified - lb_verified <= TAS_EPS)

    row['alpha_init'] = alpha_init
    row['alpha_star'] = ub_verified
    row['alpha_lb_verified'] = lb_verified
    row['alpha_lb_unverified'] = lb_unverified
    row['alpha_ub_verified'] = ub_verified
    row['verified_through_binary_search'] = verified
    row['n_bisect_iters'] = n_iters
    row['tas_time_total'] = time.time() - start_time
    results_exp2.append(row)

    print(f'[{count+1}/{len(failed_indices)}] {tag}: alpha*={ub_verified:.4f} '
          f'(lb_v={lb_verified}, lb_u={lb_unverified:.4f}, verified={verified}) '
          f'{n_iters} iters, {row["tas_time_total"]:.1f}s')

df_exp2 = pd.DataFrame(results_exp2)
if len(df_exp2) > 0:
    df_exp2.to_csv(RESULTS_DIR / 'kpg_bilinear_typeBC_exp2_tas.csv', index=False)
    print(f'\nSaved {len(df_exp2)} rows to kpg_bilinear_typeBC_exp2_tas.csv')
else:
    print('\nNo instances needed TAS (all solved in Exp1).')
```

```python
# Experiment 2 Summary
print('=== Experiment 2: TAS Summary (Type BC) ===')
if len(df_exp2) > 0:
    print(f'Instances in TAS: {len(df_exp2)}')
    print(f'Mean alpha*: {df_exp2["alpha_star"].mean():.4f}')
    print(f'Max  alpha*: {df_exp2["alpha_star"].max():.4f}')
    n_verified = df_exp2['verified_through_binary_search'].sum()
    print(f'Fully verified: {n_verified} ({n_verified/len(df_exp2):.1%})')
    n_verified_lb = df_exp2['alpha_lb_verified'].notna().sum()
    print(f'Has verified lower bound: {n_verified_lb}')
    print(f'\nPer-instance:')
    for _, r in df_exp2.iterrows():
        v = '✓' if r['verified_through_binary_search'] else '~'
        print(f'  {r["tag"]}: α*={r["alpha_star"]:.4f} '
              f'(lb_v={r["alpha_lb_verified"]}, ub_v={r["alpha_ub_verified"]:.4f}) '
              f'{v} [{r["n_bisect_iters"]} iters, {r["tas_time_total"]:.1f}s]')
else:
    print('No instances needed TAS — all solved in Exp1!')
```

```python
# Merge Exp1 and Exp2 for Type BC
if len(df_exp2) > 0:
    exp2_cols = ['tag', 'alpha_init', 'alpha_star', 'alpha_lb_verified',
                 'alpha_lb_unverified', 'alpha_ub_verified',
                 'verified_through_binary_search', 'n_bisect_iters']
    df_all_bc = pd.merge(df_exp1, df_exp2[exp2_cols], on='tag', how='left')
else:
    df_all_bc = df_exp1.copy()
    for col in ['alpha_init', 'alpha_star', 'alpha_lb_verified',
                'alpha_lb_unverified', 'alpha_ub_verified',
                'verified_through_binary_search', 'n_bisect_iters']:
        df_all_bc[col] = None

print(f'Type BC combined: {len(df_all_bc)} rows')

# Now merge with existing Type B/C results
exp1_orig = pd.read_csv(RESULTS_DIR / 'kpg_bilinear_exp1_brd_gzr_pos.csv')
exp2_orig_path = RESULTS_DIR / 'kpg_bilinear_exp2_tas.csv'
if exp2_orig_path.exists():
    exp2_orig = pd.read_csv(exp2_orig_path)
else:
    exp2_orig = pd.DataFrame()

# Combine Exp1: B + C + BC
df_exp1_all = pd.concat([exp1_orig, df_exp1], ignore_index=True)
df_exp1_all.to_csv(RESULTS_DIR / 'kpg_bilinear_exp1_brd_gzr_pos_all.csv', index=False)
print(f'Combined Exp1 (B+C+BC): {len(df_exp1_all)} rows → kpg_bilinear_exp1_brd_gzr_pos_all.csv')

# Combine Exp2: original + BC
if len(df_exp2) > 0:
    if len(exp2_orig) > 0:
        df_exp2_all = pd.concat([exp2_orig, df_exp2], ignore_index=True)
    else:
        df_exp2_all = df_exp2.copy()
    df_exp2_all.to_csv(RESULTS_DIR / 'kpg_bilinear_exp2_tas_all.csv', index=False)
    print(f'Combined Exp2 (B+C+BC): {len(df_exp2_all)} rows → kpg_bilinear_exp2_tas_all.csv')
elif len(exp2_orig) > 0:
    df_exp2_all = exp2_orig.copy()
    df_exp2_all.to_csv(RESULTS_DIR / 'kpg_bilinear_exp2_tas_all.csv', index=False)
    print(f'Combined Exp2 (B+C only): {len(df_exp2_all)} rows → kpg_bilinear_exp2_tas_all.csv')

# Final summary table
print('\n=== Combined Summary (All Types) ===')
for t in ['B', 'C', 'BC']:
    dt = df_exp1_all[df_exp1_all['interaction_type'] == t]
    n_inst = len(dt)
    n_brd = dt['brd_found_pne'].sum()
    n_gzr = (dt['gzr_status'] == 'OPTIMAL').sum()
    gzr_found = ((dt['brd_found_pne']) | (dt['gzr_status'] == 'OPTIMAL')).sum()
    pos_m = dt['pos'].dropna().mean() if dt['pos'].notna().sum() > 0 else float('nan')
    n_tl = (dt['gzr_status'] == 'TIME_LIMIT').sum()
    n_inf = (dt['gzr_status'] == 'INFEASIBLE').sum()
    print(f'  Type {t:>2}: {n_inst:3d} inst | BRD={n_brd:3d} | GZR_OPT={n_gzr:3d} | '
          f'GZR_found={gzr_found:3d} | TL={n_tl} | INF={n_inf} | mean POS={pos_m:.4f}')
```

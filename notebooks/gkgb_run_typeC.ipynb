# Bilinear KPG with Shared Constraints — Exp 1 & Exp 2

**Dataset**: Dragotto & Scatamacchia (2023) Type B and C instances  
- n ∈ {2, 3}, m ∈ {25, 50, 75, 100}, budget_ratio ∈ {2, 5, 8}  
- Type B: positive interactions C ∈ U[1, 100]  
- Type C: mixed interactions C ∈ U[-100, 100]  
- Shared item capacities c_j ~ U{1, n} (seed=42)

**Exp 1**: Best PNE and POS (BRD + GZR @ alpha=1)
- BRD warm-starts GZR; GZR optimises to MIP gap=0 (`stop_at_first=False`) for best PNE
- SO = social optimum (max welfare without equilibrium constraints)
- POS = SO / best_PNE (≥ 1)

**Exp 2**: Tightest Alpha Search (TAS) for non-PNE instances
- Uses `stop_at_first=True` (existence only, not optimality)

```python
from pathlib import Path
import sys
import time
from collections import Counter

import numpy as np
import pandas as pd

# Ensure we can import the gkg_bilinear package
    

```

```python
from gipg.gkgb.instance import KPGInstance, enumerate_dragotto_instances
from gipg.gkgb.best_response import solve_best_response
from gipg.gkgb.heuristics import brd_random_restart, alpha_of_profile
from gipg.gkgb.gzr import solve_gzr
from gipg.gkgb.abs_search import solve_abs, alpha_needed_for_profile
from gipg.gkgb.social_optimum import solve_social_optimum, compute_pos
from gipg.gkgb.objectives import social_welfare, player_utility

RESULTS_DIR = Path('../results')
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

print('Bilinear KPG modules loaded successfully.')
```

## 1. Load Instances

```python
# Path to datasets (Dragotto Type B/C copied to project)
DATA_DIR = Path('../data/gkgb')
assert DATA_DIR.exists(), f'Missing data folder: {DATA_DIR}'

SHARED_SEED = 42  # deterministic shared constraint generation

# Load Type B and C from separate subdirectories
instances_B = enumerate_dragotto_instances(DATA_DIR / 'type_B', types=('B',), shared_seed=SHARED_SEED)
instances_C = enumerate_dragotto_instances(DATA_DIR / 'type_C', types=('C',), shared_seed=SHARED_SEED)
instances = instances_B + instances_C
print(f'Loaded {len(instances)} instances ({len(instances_B)} B + {len(instances_C)} C)')

# Summary
print('\n--- By (n, m) ---')
size_counts = Counter()
for inst in instances:
    size_counts[(inst.n, inst.m)] += 1
for k in sorted(size_counts.keys()):
    print(f'  n={k[0]}, m={k[1]}: {size_counts[k]} instances')

print('\n--- By (interaction_type, budget_ratio) ---')
type_counts = Counter()
for inst in instances:
    parts = inst.tag.split('-')
    br = int(parts[2])
    type_counts[(inst.interaction_type, br)] += 1
for k in sorted(type_counts.keys()):
    print(f'  type={k[0]}, br={k[1]}: {type_counts[k]} instances')
```

## 2. Experiment 1: Best PNE and POS (BRD + GZR @ alpha=1)

For each instance:
1. **BRD** warm-starts **GZR @ alpha=1** (if BRD found a PNE)
2. **GZR** with `stop_at_first=False` — optimises to MIP gap=0, finding the **best** PNE
3. **Social Optimum** computed separately
4. **POS** = SO / best_PNE

```python
TOTAL_TIME_LIMIT = 600.0  # Total budget for BRD + GZR combined
SO_TIME_LIMIT = 600.0     # 10 minutes for social optimum
BR_TIME_LIMIT = 10.0

results_exp1 = []

for idx, inst in enumerate(instances):
    tag = inst.tag
    parts = tag.split('-')
    n_val, m_val, br_val = int(parts[0]), int(parts[1]), int(parts[2])

    row = {
        'tag': tag,
        'n': inst.n,
        'm': inst.m,
        'budget_ratio': br_val,
        'interaction_type': inst.interaction_type,
    }

    # --- Phase 1: BRD ---
    try:
        x_brd, brd_pne, brd_time = brd_random_restart(
            inst, max_init=3, max_round=15, seed=0,
            br_time_limit=BR_TIME_LIMIT,
        )
        row['brd_found_pne'] = brd_pne
        row['brd_time'] = brd_time
        row['brd_alpha'] = 1.0 if brd_pne else alpha_of_profile(inst, x_brd, br_time_limit=BR_TIME_LIMIT)
        row['initial_pne_welfare'] = social_welfare(inst, x_brd)
    except Exception as e:
        row['brd_found_pne'] = False
        row['brd_time'] = 0.0
        row['brd_alpha'] = float('inf')
        row['initial_pne_welfare'] = None
        x_brd = None
        brd_pne = False
        brd_time = 0.0
        print(f'  BRD ERROR: {e}')

    # --- Phase 2: GZR @ alpha=1, warm-started with BRD profile ---
    gzr_time_limit = max(TOTAL_TIME_LIMIT - brd_time, 60.0)
    warm = x_brd if brd_pne else None

    try:
        gzr_res = solve_gzr(
            inst, alpha=1.0, time_limit=gzr_time_limit,
            br_time_limit=BR_TIME_LIMIT,
            warm_start=warm,
            stop_at_first=False, verbose=False,
        )
        row['gzr_status'] = gzr_res.status
        row['gzr_mip_gap'] = gzr_res.mip_gap
        row['gzr_obj_bound'] = gzr_res.obj_bound
        row['gzr_cuts'] = gzr_res.cuts_added
        row['gzr_br_calls'] = gzr_res.br_calls
        row['gzr_time'] = brd_time + gzr_res.runtime
        row['gzr_obj_val'] = gzr_res.obj_val
        row['gzr_first_pne_time'] = (
            brd_time + gzr_res.first_pne_time
            if gzr_res.first_pne_time is not None else None
        )
    except Exception as e:
        gzr_res = None
        row['gzr_status'] = 'ERROR'
        row['gzr_mip_gap'] = None
        row['gzr_obj_bound'] = None
        row['gzr_cuts'] = None
        row['gzr_br_calls'] = None
        row['gzr_time'] = None
        row['gzr_obj_val'] = None
        row['gzr_first_pne_time'] = None
        print(f'  GZR ERROR: {e}')

    # Best PNE welfare (using full bilinear utility)
    if gzr_res is not None and gzr_res.profile is not None and gzr_res.status in ('OPTIMAL', 'FEASIBLE'):
        row['best_pne_welfare'] = social_welfare(inst, gzr_res.profile)
    elif brd_pne:
        row['best_pne_welfare'] = row.get('initial_pne_welfare')
    else:
        row['best_pne_welfare'] = None

    # First PNE time
    row['first_pne_time'] = row.get('gzr_first_pne_time')

    # --- Phase 3: Social Optimum ---
    try:
        so_res = solve_social_optimum(inst, time_limit=SO_TIME_LIMIT, verbose=False)
        row['so_status'] = so_res.status
        row['so_welfare'] = so_res.opt_cost
        row['so_time'] = so_res.runtime
    except Exception as e:
        row['so_status'] = 'ERROR'
        row['so_welfare'] = None
        row['so_time'] = None
        print(f'  SO ERROR: {e}')

    # --- POS ---
    row['pos'] = None
    if (row.get('best_pne_welfare') is not None
            and row.get('so_welfare') is not None
            and row['best_pne_welfare'] > 0):
        row['pos'] = compute_pos(row['so_welfare'], row['best_pne_welfare'])

    results_exp1.append(row)

    # Progress
    pne_str = 'BRD-PNE' if brd_pne else 'no-PNE'
    gzr_time_str = f'{row["gzr_time"]:.1f}s' if row.get('gzr_time') is not None else '?s'
    gzr_cuts_str = f'{row["gzr_cuts"]}cuts' if row.get('gzr_cuts') is not None else '?cuts'
    pos_str = f'POS={row["pos"]:.3f}' if row['pos'] is not None else 'POS=N/A'
    fpne_str = f'1stPNE={row["gzr_first_pne_time"]:.1f}s' if row.get('gzr_first_pne_time') is not None else '1stPNE=N/A'
    if (idx + 1) % 10 == 0 or idx == len(instances) - 1:
        print(f'[{idx+1}/{len(instances)}] {tag}: {pne_str} | GZR {row.get("gzr_status","?")} '
              f'({gzr_time_str}, {gzr_cuts_str}) | {pos_str} | {fpne_str}')

df_exp1 = pd.DataFrame(results_exp1)
df_exp1.to_csv(RESULTS_DIR / 'kpg_bilinear_exp1_brd_gzr_pos.csv', index=False)
print(f'\nSaved {len(df_exp1)} rows to kpg_bilinear_exp1_brd_gzr_pos.csv')
```

```python
# Experiment 1 Summary
print('=== Experiment 1 Summary ===')

n_total = len(df_exp1)
n_brd_pne = df_exp1['brd_found_pne'].sum()
n_gzr_opt = (df_exp1['gzr_status'] == 'OPTIMAL').sum()
n_gzr_feas = (df_exp1['gzr_status'] == 'FEASIBLE').sum()
n_gzr_inf = (df_exp1['gzr_status'] == 'INFEASIBLE').sum()
n_gzr_tl = (df_exp1['gzr_status'] == 'TIME_LIMIT').sum()
n_pos = df_exp1['pos'].notna().sum()

print(f'Total instances: {n_total}')
print(f'BRD found PNE: {n_brd_pne} ({n_brd_pne/n_total:.1%})')
print(f'GZR OPTIMAL: {n_gzr_opt} ({n_gzr_opt/n_total:.1%})')
print(f'GZR FEASIBLE: {n_gzr_feas}')
print(f'GZR INFEASIBLE: {n_gzr_inf} ({n_gzr_inf/n_total:.1%})')
print(f'GZR TIME_LIMIT: {n_gzr_tl} ({n_gzr_tl/n_total:.1%})')
print(f'POS computed: {n_pos} ({n_pos/n_total:.1%})')

# By (n, m)
print('\n--- By (n, m) ---')
summary = df_exp1.groupby(['n', 'm']).agg(
    count=('tag', 'count'),
    brd_pne_rate=('brd_found_pne', 'mean'),
    gzr_opt_rate=('gzr_status', lambda x: (x == 'OPTIMAL').mean()),
    avg_gzr_time=('gzr_time', 'mean'),
    avg_pos=('pos', 'mean'),
).round(4)
print(summary)

# By (interaction_type, budget_ratio)
print('\n--- By (interaction_type, budget_ratio) ---')
summary_type = df_exp1.groupby(['interaction_type', 'budget_ratio']).agg(
    count=('tag', 'count'),
    brd_pne_rate=('brd_found_pne', 'mean'),
    gzr_opt_rate=('gzr_status', lambda x: (x == 'OPTIMAL').mean()),
    avg_gzr_time=('gzr_time', 'mean'),
    avg_pos=('pos', 'mean'),
).round(4)
print(summary_type)

# By (n, interaction_type)
print('\n--- By (n, interaction_type) ---')
summary_ni = df_exp1.groupby(['n', 'interaction_type']).agg(
    count=('tag', 'count'),
    brd_pne_rate=('brd_found_pne', 'mean'),
    gzr_opt_rate=('gzr_status', lambda x: (x == 'OPTIMAL').mean()),
    avg_gzr_cuts=('gzr_cuts', 'mean'),
    avg_gzr_time=('gzr_time', 'mean'),
    avg_pos=('pos', 'mean'),
).round(4)
print(summary_ni)
```

## 3. Experiment 2: Tightest Alpha Search (TAS) for Non-PNE Instances

Only instances where Exp 1 GZR reported **INFEASIBLE** or **TIME_LIMIT** (and BRD found no PNE).  
Bisection with verified/unverified bounds.

```python
df_exp1 = pd.read_csv(RESULTS_DIR / 'kpg_bilinear_exp1_brd_gzr_pos.csv')
results_exp1 = df_exp1.to_dict('records')
```

```python
BR_TIME_LIMIT = 10.0
TAS_GZR_LIMIT = 1800.0   # 20 minutes per GZR call in TAS
TAS_EPS = 1e-2
TAS_MAX_ITERS = 30

# Gather failed instances from Exp1
failed_indices = [
    idx for idx, r in enumerate(results_exp1)
    if r['gzr_status'] in ('INFEASIBLE', 'TIME_LIMIT') and not r.get('brd_found_pne', False)
]
print(f'Exp2 candidates: {len(failed_indices)} instances')

results_exp2 = []

for count, idx in enumerate(failed_indices):
    inst = instances[idx]
    tag = inst.tag
    start_time = time.time()

    row = {
        'tag': tag,
        'n': inst.n,
        'm': inst.m,
        'budget_ratio': int(tag.split('-')[2]),
        'interaction_type': inst.interaction_type,
    }

    # --- Phase 0: BRD upper bound on alpha ---
    x_brd, brd_pne, _ = brd_random_restart(
        inst, max_init=5, max_round=20, seed=0,
        br_time_limit=BR_TIME_LIMIT,
    )
    alpha_init = 1.0 if brd_pne else alpha_of_profile(inst, x_brd, br_time_limit=BR_TIME_LIMIT)
    ub_verified = alpha_init
    best_profile = x_brd

    # --- Phase 1: Reuse Exp1 result for alpha=1 ---
    exp1_row = results_exp1[idx]
    if exp1_row['gzr_status'] == 'INFEASIBLE':
        lb_verified = 1.0
        lb_unverified = 1.0
    else:  # TIME_LIMIT
        lb_verified = None
        lb_unverified = 1.0

    # --- Phase 2: Bisection ---
    n_iters = 0
    history = []

    while (ub_verified - lb_unverified > TAS_EPS) and (n_iters < TAS_MAX_ITERS):
        n_iters += 1
        alpha_mid = 0.5 * (lb_unverified + ub_verified)

        gzr_mid = solve_gzr(
            inst, alpha=alpha_mid, time_limit=TAS_GZR_LIMIT,
            br_time_limit=BR_TIME_LIMIT,
            stop_at_first=True, verbose=False,
        )

        if gzr_mid.status in ('FEASIBLE', 'OPTIMAL') and gzr_mid.profile is not None:
            ub_verified = alpha_mid
            best_profile = gzr_mid.profile
            bound_type = 'verified_ub'
        elif gzr_mid.status == 'INFEASIBLE':
            lb_unverified = alpha_mid
            lb_verified = alpha_mid if lb_verified is None else max(lb_verified, alpha_mid)
            bound_type = 'verified_lb'
        else:
            lb_unverified = alpha_mid
            bound_type = 'unverified_lb'

        history.append({
            'iter': n_iters, 'alpha_mid': alpha_mid,
            'status': gzr_mid.status, 'bound_type': bound_type,
            'runtime': gzr_mid.runtime,
        })
        print(f'  iter {n_iters}: alpha={alpha_mid:.4f} -> {gzr_mid.status} ({bound_type}) '
              f'[{gzr_mid.runtime:.1f}s]')

    verified = (lb_verified is not None) and (ub_verified - lb_verified <= TAS_EPS)

    row['alpha_init'] = alpha_init
    row['alpha_star'] = ub_verified
    row['alpha_lb_verified'] = lb_verified
    row['alpha_lb_unverified'] = lb_unverified
    row['alpha_ub_verified'] = ub_verified
    row['verified_through_binary_search'] = verified
    row['n_bisect_iters'] = n_iters
    row['tas_time_total'] = time.time() - start_time
    results_exp2.append(row)

    print(f'[{count+1}/{len(failed_indices)}] {tag}: alpha*={ub_verified:.4f} '
          f'(lb_v={lb_verified}, lb_u={lb_unverified:.4f}, verified={verified}) '
          f'{n_iters} iters, {row["tas_time_total"]:.1f}s')

df_exp2 = pd.DataFrame(results_exp2)
if len(df_exp2) > 0:
    df_exp2.to_csv(RESULTS_DIR / 'kpg_bilinear_exp2_tas.csv', index=False)
    print(f'\nSaved {len(df_exp2)} rows to kpg_bilinear_exp2_tas.csv')
else:
    print('\nNo instances needed TAS (all solved in Exp1).')
```

```python
# Experiment 2 Summary
print('=== Experiment 2: TAS Summary ===')
if len(df_exp2) > 0:
    print(f'Instances in TAS: {len(df_exp2)}')
    print(f'Mean alpha*: {df_exp2["alpha_star"].mean():.4f}')
    print(f'Max  alpha*: {df_exp2["alpha_star"].max():.4f}')
    n_verified = df_exp2['verified_through_binary_search'].sum()
    print(f'Fully verified: {n_verified} ({n_verified/len(df_exp2):.1%})')
    n_verified_lb = df_exp2['alpha_lb_verified'].notna().sum()
    print(f'Has verified lower bound: {n_verified_lb} ({n_verified_lb/len(df_exp2):.1%})')
    print(f'Mean bisection iters: {df_exp2["n_bisect_iters"].mean():.1f}')

    print('\n--- By (n, interaction_type) ---')
    summary2 = df_exp2.groupby(['n', 'interaction_type']).agg(
        count=('tag', 'count'),
        avg_alpha_star=('alpha_star', 'mean'),
        max_alpha_star=('alpha_star', 'max'),
        avg_iters=('n_bisect_iters', 'mean'),
        n_verified=('verified_through_binary_search', 'sum'),
    ).round(4)
    print(summary2)

    print('\n--- By (n, m) ---')
    summary2_nm = df_exp2.groupby(['n', 'm']).agg(
        count=('tag', 'count'),
        avg_alpha_star=('alpha_star', 'mean'),
        max_alpha_star=('alpha_star', 'max'),
        avg_iters=('n_bisect_iters', 'mean'),
    ).round(4)
    print(summary2_nm)
else:
    print('No instances required TAS.')
```

## 4. Combined Results

```python
# Merge Exp1 and Exp2
if len(df_exp2) > 0:
    exp2_cols = ['tag', 'alpha_init', 'alpha_star', 'alpha_lb_verified',
                 'alpha_lb_unverified', 'alpha_ub_verified',
                 'verified_through_binary_search', 'n_bisect_iters']
    df_all = pd.merge(df_exp1, df_exp2[exp2_cols], on='tag', how='left')
else:
    df_all = df_exp1.copy()
    for col in ['alpha_init', 'alpha_star', 'alpha_lb_verified',
                'alpha_lb_unverified', 'alpha_ub_verified',
                'verified_through_binary_search', 'n_bisect_iters']:
        df_all[col] = np.nan

# For instances solved in Exp1, alpha_star = 1.0
mask_solved = (
    (df_all['gzr_status'] == 'OPTIMAL') |
    (df_all['gzr_status'] == 'FEASIBLE') |
    (df_all['brd_found_pne'] == True)
)
df_all.loc[mask_solved & df_all['alpha_star'].isna(), 'alpha_star'] = 1.0

df_all.to_csv(RESULTS_DIR / 'kpg_bilinear_results_all.csv', index=False)
print(f'Combined results: {len(df_all)} rows -> kpg_bilinear_results_all.csv')
print(f'Columns: {list(df_all.columns)}')

# Final summary
print('\n=== Final Summary ===')
for n_val in sorted(df_all['n'].unique()):
    for itype in sorted(df_all['interaction_type'].unique()):
        sub = df_all[(df_all['n'] == n_val) & (df_all['interaction_type'] == itype)]
        n_pne = (
            (sub['gzr_status'].isin(['OPTIMAL', 'FEASIBLE'])) | (sub['brd_found_pne'] == True)
        ).sum()
        pos_ok = sub['pos'].notna()
        pos_str = f'POS mean={sub.loc[pos_ok, "pos"].mean():.3f}' if pos_ok.any() else 'no POS'
        tas_sub = sub[sub['n_bisect_iters'].notna() & (sub['n_bisect_iters'] > 0)]
        tas_str = f'TAS: {len(tas_sub)} inst' if len(tas_sub) > 0 else 'no TAS'
        print(f'  n={n_val}, type={itype}: {len(sub)} inst, PNE={n_pne}, {pos_str}, {tas_str}')
```
